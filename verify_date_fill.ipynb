After we fill in the station names for the dates we were missing by propogating the last valid observation forward,
verify that none of the missing dates were on January 1 2013, which would put us at risk of of propogating the wrong station name forward.

```python
import pandas as pd
```

```python
data = pd.read_csv("test.csv")
```

```python
firstDate = []
# loop through all of the stations and see if they're missing the January 1st 2013, if not we can simply
# fill in the missing values with the last valid observation without worrying about incorrectly
# assigning a station name
for station in data["STATION_NAME"].unique():
    if station == 'nan':
        continue
    else:
        if data[data["STATION_NAME"] == station]["LOCAL_DATE"].unique()[0] == '2013-01-01':
            firstDate.append(station)
```

```python
assert len(firstDate) == len(data["STATION_NAME"].unique())
```
